**SedenionSpectralRelativity companion notebook.** First notebook in this

The recipe, generic: one SCALE-type generator per basis position, composed

(GenerationalLineage) runs it on plain integers under `a*b`;
`bracket_firing_engine.py` (this repo) runs it on the sedenion's own `L_a`

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from bracket_firing_engine import build_tree, firing_order, verify_block_identity, spectral_summary
from emerger_spectrum import text_to_vec, parse_vec

vec = text_to_vec("primes are words")
ok = verify_block_identity(vec)
print("block identity verified on this input, every level:", ok)
```

## The 31-node bifurcation tree, firing order

Sedenion (dim 16) -> two orthogonal octonions (dim 8) -> each's two
orthogonal quaternions (dim 4) -> complex pairs (dim 2) -> reals (dim 1).
31 nodes. The firing order is a one-way depth-first cursor: smaller-norm
child fires first.

```python
tree = build_tree(vec)
order = firing_order(tree)
print(f"{len(order)} nodes, firing order:")
print("  " + "  ".join(f'{n["path"]}({n["layer"]}{n["dim"]})' for n in order))
```

## The spectral texture — Hurwitz, checked directly

Every NONZERO node below dim 16 is provably an exact isometry (Hurwitz:
R, C, H, O are the only normed division algebras). The sedenion root is
the only place eigenvalue-modulus can spread — that departure IS the
zero-divisor fault, read directly off the spectrum.

```python
summ = spectral_summary(tree)
for dim in (16, 8, 4, 2, 1):
    s = summ[dim]
    print(f"dim {dim:>2}: nodes={s['n_nodes']:<2} zero-divisors={s['n_zd']:<2} "
          f"isometry_holds={s['per_node_spread_all_one']}")
```
